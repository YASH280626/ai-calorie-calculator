```python
import sqlite3
```

```python
csv_path = '../data/processed/Indian_Food_Nutrition_Processed.csv'
db_path = '../data/processed/food_nutrition.db'
```

```python
clean_df = pd.read_csv(csv_path)

clean_df.head()
```

```python
conn = sqlite3.connect(db_path)

clean_df.to_sql(
    name='food_nutrition',
    con=conn,
    if_exists='replace',
    index=False
)

print("Database created successfully!")
```

```python
cursor = conn.cursor()

cursor.execute("SELECT COUNT(*) FROM food_nutrition")

print("Total dishes:", cursor.fetchone()[0])
```

```python
result = pd.read_sql_query(
    "SELECT * FROM food_nutrition LIMIT 5",
    conn
)

result
```
